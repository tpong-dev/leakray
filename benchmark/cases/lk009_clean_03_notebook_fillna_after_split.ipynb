```python
# expect: clean
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

## Setup


```python
%matplotlib inline
```

```python
df = pd.read_csv("weather_stations.csv").drop_duplicates()
df["rain_mm"].hist()
plt.show()
```

```python
train, test = train_test_split(df, test_size=0.2, random_state=0)
fill = train.mean(numeric_only=True)
train, test = train.fillna(fill), test.fillna(fill)
LinearRegression().fit(train[["pressure"]], train["rain_mm"]).score(test[["pressure"]], test["rain_mm"])
```
