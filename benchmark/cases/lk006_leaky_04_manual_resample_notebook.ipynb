```python
# expect: LK006
import pandas as pd
from sklearn.utils import resample
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
```

## Setup


```python
%matplotlib inline
```

```python
df = pd.read_csv("claims.csv").drop_duplicates()
majority = df[df["approved"] == 0]
minority = df[df["approved"] == 1]
minority_up = resample(minority, replace=True, n_samples=len(majority), random_state=1)
balanced = pd.concat([majority, minority_up])
```

```python
X_train, X_test, y_train, y_test = train_test_split(balanced.drop(columns="approved"), balanced["approved"])
gb = GradientBoostingClassifier().fit(X_train, y_train)
gb.score(X_test, y_test)
```
