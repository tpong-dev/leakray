```python
# expect: clean
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
```

## Setup


```python
%matplotlib inline
```

```python
df = pd.read_csv("hepatitis.csv").drop_duplicates()
pipe = Pipeline([("imp", SimpleImputer()), ("rf", RandomForestClassifier())])
gs = GridSearchCV(pipe, {"rf__max_depth": [3, 5, None], "imp__strategy": ["mean", "median"]}, cv=5)
```

```python
gs.fit(df.drop(columns="died"), df["died"])
gs.best_score_
```
